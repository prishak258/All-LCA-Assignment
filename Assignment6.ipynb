```python
import pandas as pd

# Step 1: Create a sample dataset
data = {
    'Name': ['Prisha', 'Aarav', 'Riya', 'Kabir', 'Prisha'],
    'Marks': [85, 90, None, 75, 85],
    'City': ['Pune', 'Mumbai', 'Delhi', None, 'Pune'],
    'Age': [18, 19, 18, 20, 18],
    'Remarks': ['Good', 'Excellent', 'Average', 'Good', 'Good']
}

df = pd.DataFrame(data)

# Display original dataset
print("Original Dataset:")
print(df)

# Step 2: Explore the dataset
print("\nFirst Five Rows:")
print(df.head())

print("\nDataset Information:")
df.info()

print("\nStatistical Description:")
print(df.describe())

# Step 3: Check for missing values
print("\nMissing Values:")
print(df.isnull().sum())

# Step 4: Handle missing values
df['Marks'] = df['Marks'].fillna(df['Marks'].mean())
df['City'] = df['City'].fillna('Unknown')

print("\nDataset After Handling Missing Values:")
print(df)

# Step 5: Remove duplicate records
df = df.drop_duplicates()

print("\nDataset After Removing Duplicates:")
print(df)

# Step 6: Drop an irrelevant column
df = df.drop(columns=['Remarks'])

print("\nDataset After Dropping Irrelevant Column:")
print(df)

# Step 7: Select specific rows and columns using loc[]
print("\nStudents with Marks Greater Than 80:")
print(df.loc[df['Marks'] > 80, ['Name', 'Marks', 'City']])

# Step 8: Filter students from Pune
print("\nStudents from Pune:")
print(df.loc[df['City'] == 'Pune'])

# Step 9: Save the preprocessed dataset
df.to_csv('preprocessed_data.csv', index=False)

print("\nPreprocessed dataset saved successfully!")
```
