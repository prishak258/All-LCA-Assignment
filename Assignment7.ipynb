```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Step 1: Create a sample dataset
data = {
    'Student': ['A', 'B', 'C', 'D', 'E'],
    'Marks': [85, 90, 75, 95, 80],
    'Attendance': [90, 85, 80, 95, 75],
    'Study_Hours': [4, 5, 3, 6, 2]
}

df = pd.DataFrame(data)

# Step 2: Display and explore the dataset
print("Dataset:")
print(df)

print("\nFirst Five Rows:")
print(df.head())

print("\nDataset Information:")
df.info()

print("\nStatistical Description:")
print(df.describe())

# Step 3: Create a figure with four plots
plt.figure(figsize=(12, 9))

# Line Plot
plt.subplot(2, 2, 1)
plt.plot(df['Student'], df['Marks'], marker='o', label='Marks')
plt.title('Line Plot - Student Marks')
plt.xlabel('Student')
plt.ylabel('Marks')
plt.legend()
plt.grid(True)

# Bar Plot
plt.subplot(2, 2, 2)
plt.bar(df['Student'], df['Attendance'], label='Attendance')
plt.title('Bar Plot - Student Attendance')
plt.xlabel('Student')
plt.ylabel('Attendance (%)')
plt.legend()

# Scatter Plot
plt.subplot(2, 2, 3)
plt.scatter(df['Study_Hours'], df['Marks'], label='Students')
plt.title('Scatter Plot - Study Hours vs Marks')
plt.xlabel('Study Hours')
plt.ylabel('Marks')
plt.legend()
plt.grid(True)

# Heatmap
plt.subplot(2, 2, 4)
correlation = df[['Marks', 'Attendance', 'Study_Hours']].corr()
sns.heatmap(correlation, annot=True, cmap='coolwarm', fmt='.2f')
plt.title('Heatmap - Correlation Matrix')

# Step 4: Display all plots
plt.tight_layout()
plt.show()
```
