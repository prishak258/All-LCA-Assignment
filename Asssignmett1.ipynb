```python
# Assignment No. 1
# Different Operations on List, Tuple and Dictionary

# ---------------- LIST ----------------

print("----- LIST OPERATIONS -----")

students_list = ["Aarav", "Priya", "Rahul", "Sneha"]

print("Original List:", students_list)

# Add
students_list.append("Rohan")
print("After Adding:", students_list)

# Update
students_list[1] = "Ananya"
print("After Updating:", students_list)

# Delete
students_list.remove("Rahul")
print("After Deleting:", students_list)


# ---------------- TUPLE ----------------

print("\n----- TUPLE OPERATIONS -----")

students_tuple = ("Aarav", "Priya", "Rahul", "Sneha")

print("Original Tuple:", students_tuple)

# Add
# Tuples are immutable, so we create a new tuple
students_tuple = students_tuple + ("Rohan",)
print("After Adding:", students_tuple)

# Update
# Tuples cannot be directly updated, so convert to list
temp_list = list(students_tuple)
temp_list[1] = "Ananya"
students_tuple = tuple(temp_list)
print("After Updating:", students_tuple)

# Delete
# Tuples cannot directly delete an element, so convert to list
temp_list = list(students_tuple)
temp_list.remove("Rahul")
students_tuple = tuple(temp_list)
print("After Deleting:", students_tuple)


# ---------------- DICTIONARY ----------------

print("\n----- DICTIONARY OPERATIONS -----")

students_dict = {
    101: "Aarav",
    102: "Priya",
    103: "Rahul",
    104: "Sneha"
}

print("Original Dictionary:", students_dict)

# Add
students_dict[105] = "Rohan"
print("After Adding:", students_dict)

# Update
students_dict[102] = "Ananya"
print("After Updating:", students_dict)

# Delete
del students_dict[103]
print("After Deleting:", students_dict)
```
