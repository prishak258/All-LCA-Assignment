```python
# Enter the order of matrices
rows = int(input("Enter number of rows: "))
cols = int(input("Enter number of columns: "))

# Create first matrix
print("\nEnter elements of Matrix A:")
matrix_a = []

for i in range(rows):
    row = []
    for j in range(cols):
        value = int(input(f"Enter A[{i}][{j}]: "))
        row.append(value)
    matrix_a.append(row)

# Create second matrix
print("\nEnter elements of Matrix B:")
matrix_b = []

for i in range(rows):
    row = []
    for j in range(cols):
        value = int(input(f"Enter B[{i}][{j}]: "))
        row.append(value)
    matrix_b.append(row)

# Add the two matrices
matrix_sum = []

for i in range(rows):
    row = []
    for j in range(cols):
        row.append(matrix_a[i][j] + matrix_b[i][j])
    matrix_sum.append(row)

# Display matrices
print("\nMatrix A:")
for row in matrix_a:
    print(row)

print("\nMatrix B:")
for row in matrix_b:
    print(row)

print("\nAddition of Matrix A and Matrix B:")
for row in matrix_sum:
    print(row)
```
